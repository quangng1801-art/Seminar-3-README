## 1. Importieren von Bibliotheken

Mit `import` bindest du vorhandene Bibliotheken ein. `math` bietet mathematische Funktionen; `random` erzeugt Zufallswerte.

```python
import math
import random

print("Quadratwurzel aus 81:", math.sqrt(81))
print("Zufallszahl zwischen 1 und 6:", random.randint(1, 6))
```

## 2. Variablen und Datentypen

Variablen geben Werten einen Namen. Häufige Grundtypen sind `str` (Text), `int` (Ganzzahl), `float` (Kommazahl) und `bool` (Wahrheitswert).

```python
name = "Mina"
alter = 21
koerpergroesse = 1.68
studentin = True

print(name, type(name))
print(alter, type(alter))
print(koerpergroesse, type(koerpergroesse))
print(studentin, type(studentin))
```

## 3. Operatoren und Ausdrücke

Arithmetische Operatoren berechnen Werte. Vergleichsoperatoren liefern `True` oder `False`; logische Operatoren wie `and`, `or` und `not` verknüpfen Wahrheitswerte.

```python
preis = 12.50
anzahl = 3
summe = preis * anzahl

print("Gesamt:", summe)
print("Ist der Gesamtpreis mindestens 30?", summe >= 30)
print("Gibt es mehr als 1 Artikel und kostet einer unter 20?", anzahl > 1 and preis < 20)
print("Ist die Summe nicht gleich 0?", not summe == 0)
```

## 4. Kontrollstrukturen: `if`, `elif`, `else`

Mit Bedingungen steuerst du, welcher Code ausgeführt wird. Python prüft die Zweige der Reihe nach und führt den ersten passenden aus.

```python
punktzahl = 84

if punktzahl >= 90:
    note = "Sehr gut"
elif punktzahl >= 75:
    note = "Gut"
else:
    note = "Weiter üben"

print(f"Bei {punktzahl} Punkten lautet die Bewertung: {note}.")
```

## 5. Schleifen: `for` und `while`

Eine `for`-Schleife durchläuft Elemente einer Folge. Eine `while`-Schleife läuft, solange ihre Bedingung wahr ist. Achte bei `while` darauf, die Bedingung im Schleifenkörper irgendwann zu verändern.

```python
for zahl in range(1, 4):
    print("for-Runde:", zahl)

restzeit = 3
while restzeit > 0:
    print("Noch", restzeit)
    restzeit -= 1
```

## 6. Funktionen definieren und aufrufen

Mit `def` definierst du eine Funktion. Parameter nehmen Eingaben entgegen; `return` gibt ein Ergebnis zurück.

```python
def begruessung(name):
    return f"Hallo, {name}!"

nachricht = begruessung("Mina")
print(nachricht)
```

## 7. Listen und Dictionaries

Listen speichern mehrere Werte in einer Reihenfolge. Dictionaries ordnen Werten Schlüssel zu. Beide Sammlungen lassen sich bearbeiten.

```python
farben = ["blau", "grün", "rot"]
print("Erstes Element:", farben[0])
farben.append("gelb")
print("Liste nach append:", farben)

person = {"name": "Mina", "alter": 21}
print("Name:", person["name"])
person["stadt"] = "Berlin"
print("Dictionary nach Änderung:", person)
```
